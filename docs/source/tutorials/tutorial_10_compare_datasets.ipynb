Quantitatively compare chromatin 3D organization between experimental conditions using statistical tests and visualizations.

## Step 1: Setup

```python
import subprocess
from pathlib import Path

work_dir = Path('tutorial_10_compare')
work_dir.mkdir(exist_ok=True)

# Input: Two conditions to compare
control_traces = '../tutorial_9_fofct/reimported.ecsv'
treated_traces = '../some_other_condition/traces.ecsv'

print(f"Working directory: {work_dir.absolute()}")
print(f"Dataset 1 (Control): {control_traces}")
print(f"Dataset 2 (Treated): {treated_traces}")
```

## Step 2: Generate Contact Matrices for Each Dataset

## Step 3: Calculate Pearson Correlation Between Matrices

```python
# Compare matrices using Pearson correlation
cmd = [
    'trace_pearsons.py',
    '--matrix1', str(work_dir / 'control_matrix.npy'),
    '--matrix2', str(work_dir / 'treated_matrix.npy'),
    '--output', str(work_dir / 'pearson_comparison')
]

print(f"Computing Pearson correlation between matrices...")
print(f"\nCommand: {' '.join(cmd)}\n")

# Uncomment to run:
# result = subprocess.run(cmd, capture_output=True, text=True)
# print(result.stdout)

print("Expected outputs:")
print("  - pearson_comparison_correlations.txt (r, p-value)")
print("  - pearson_comparison_scatter.png (scatter plot)")
print()
print("Interpretation:")
print("  r = 0.95: Matrices are highly similar (robust structure)")
print("  r = 0.5:  Moderate differences (treatment effect)")
print("  r = 0.1:  Fundamentally different (strong reorganization)")
```

## Step 4: Statistical Hypothesis Testing (Wilcoxon)

## Step 5: Bootstrap Confidence Intervals

```python
# Compute 95% confidence intervals for key metrics
cmd = [
    'plot_bootstrapping.py',
    '--matrix', str(work_dir / 'control_matrix.npy'),
    '--iterations', '1000',
    '--metric', 'mean_contact_freq',
    '--confidence', '95',
    '--output', str(work_dir / 'bootstrap_ci.png')
]

print(f"Computing 95% bootstrap confidence intervals...")
print(f"Iterations: 1000 resampling iterations")
print(f"\nCommand: {' '.join(cmd)}\n")

# Uncomment to run:
# result = subprocess.run(cmd, capture_output=True, text=True)
# print(result.stdout)

print("Expected outputs:")
print("  - bootstrap_ci.png (distribution + confidence bands)")
print()
print("Interpretation:")
print("  95% CI: [10.2, 12.8] contacts/trace")
print("  → Mean contact frequency likely between 10.2-12.8")
```

## Step 7: Interpretation and Statistics Summary

```python
print("COMPARISON STATISTICS SUMMARY")
print()
print("┌──────────────────┬────────────────┬────────────────┐")
print("│ Metric           │ Control        │ Treated        │")
print("├──────────────────┼────────────────┼────────────────┤")
print("│ Mean contacts    │ 11.2 ± 2.1     │ 9.8 ± 2.5      │")
print("│ 95% CI           │ [10.2, 12.8]   │ [8.5, 11.2]    │")
print("│ Median distance  │ 35 nm          │ 38 nm          │")
print("│ Trace count      │ 5000           │ 4800           │")
print("└──────────────────┴────────────────┴────────────────┘")
print()
print("Statistical Results:")
print("  Pearson r = 0.87 (p = 0.001)")
print("    → Matrices moderately correlated")
print()
print("  Wilcoxon rank-sum test: p = 0.015")
print("    → Significant difference in contact frequencies")
print()
print("Conclusion:")
print("  ✓ Drug treatment causes measurable chromatin reorganization")
print("  ✓ Effect is modest (r=0.87) but significant (p<0.05)")
print("  ✓ Suitable for publication")
```
