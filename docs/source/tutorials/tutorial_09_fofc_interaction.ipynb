# Tutorial 9: Export/Import Traces in FOF-CT Standard Format

## Objective
Convert traceratops ECSV format to/from the 4DN FISH Omics Format - Chromatin Tracing (FOF-CT) standard for data sharing and publication.

## Scientific Context
The [4DN Data Portal](https://www.4dnucleome.org/) requires FOF-CT format for:
- Public data submission
- Cross-tool interoperability
- Genomic coordinate tracking
- Reproducibility and open science

FOF-CT stores:
- Barcode sequences with genomic positions (BED format)
- 3D coordinates for each barcode per trace
- Metadata (cell type, experimental conditions)

## Scripts Used
- `trace_export_to_fofct.py` - Convert ECSV → FOF-CT (with BED mapping)
- `trace_import_from_fofct.py` - Convert FOF-CT → ECSV (with coordinate parsing)
