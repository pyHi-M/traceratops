# Tutorial 6: Assign Spatial Masks and Split by Labels

## Objective
Assign labels to traces based on spatial regions (masks), then create labeled subsets for separate analysis.

## Scientific Context
Experimental samples often have multiple regions of interest (ROIs):
- Different cell types
- Different chromosomes
- Control vs. experimental conditions
- Nuclear vs. cytoplasmic regions

Masks define spatial boundaries for each region, and labels identify which traces belong where.

## Scripts Used
- `trace_assign_mask.py` - Assign trace coordinates to regions (label assignment)
- `trace_split_labels.py` - ⭐ **NEW SCRIPT** to split by label (keep or remove)

## Step 1: Setup

````python
import subprocess
from pathlib import Path

work_dir = Path('tutorial_6_assign_masks')
work_dir.mkdir(exist_ok=True)

# Input from tutorial 5
input_trace = Path('../tutorial_5_split_traces/split_*.ecsv')

print(f"Working directory: {work_dir.absolute()}")
print(f"Input trace file: {input_trace}")
print()
print("🆕 Using NEW script: trace_split_labels.py")
````

## Step 2: Prepare Mask File (Optional - if not already created)

A mask file defines spatial regions. Example structure (JSON):

```json
{
  "region_A": {"x_min": 0, "x_max": 100, "y_min": 0, "y_max": 50},
  "region_B": {"x_min": 100, "x_max": 200, "y_min": 0, "y_max": 50},
  "region_C": {"x_min": 50, "x_max": 150, "y_min": 50, "y_max": 100}
}
```

## Step 3: Assign Labels Based on Mask Coordinates

````python
# Assign mask labels to traces based on their coordinates
cmd = [
    'trace_assign_mask.py',
    '--input', str(input_trace),
    '--maskfile', 'regions.json',
    '--output', str(work_dir / 'labeled')
]

print(f"Assigning spatial masks to traces...")
print(f"\nCommand: {' '.join(cmd)}\n")

# Uncomment to run:
# result = subprocess.run(cmd, capture_output=True, text=True)
# print(result.stdout)

print("Expected outputs:")
print("  - labeled_*.ecsv (traces with new 'label' column)")
print()
print("Each trace now has a label (e.g., 'region_A', 'region_B')")
print("indicating which spatial region it belongs to.")
````

## Step 5: REMOVE Mode - Exclude Specific Regions

````python
# Remove traces from specific region (e.g., background, artifacts)
cmd = [
    'trace_split_labels.py',
    '--input', str(work_dir / 'labeled_*.ecsv'),
    '--remove-label', 'background',
    '--output', str(work_dir / 'no_background'),
    '--stats'
]

print(f"Creating filtered subset (REMOVE mode)...")
print(f"Removing: traces labeled 'background'")
print(f"\nCommand: {' '.join(cmd)}\n")

# Uncomment to run:
# result = subprocess.run(cmd, capture_output=True, text=True)
# print(result.stdout)

print("Expected outputs:")
print("  - no_background_remove_background.ecsv")
print("  - no_background_remove_background_stats.txt")
````

## Step 6: Create Multiple Subsets for Comparison

````python
# Create subsets for each major region
regions = ['region_A', 'region_B', 'region_C']

for region in regions:
    cmd = [
        'trace_split_labels.py',
        '--input', str(work_dir / 'labeled_*.ecsv'),
        '--keep-label', region,
        '--output', str(work_dir / f'{region}_subset'),
        '--stats'
    ]
    
    print(f"Creating subset for {region}...")
    print(f"Command: {' '.join(cmd)}")
    print()
    
    # Uncomment to run all:
    # result = subprocess.run(cmd, capture_output=True, text=True)
    # print(result.stdout)
````

## Step 7: Interpret Statistics Output

````python
print("Statistics files (*_stats.txt) contain:")
print()
print("  Input count:    Total traces before splitting")
print("  Output count:   Traces kept (or removed)")
print("  Retention:      Percentage of data kept")
print()
print("Example:")
print("  Input count:  5000")
print("  Output count: 1250")
print("  Retention:    25.0%")
print()
print("← Region represents 1/4 of all traces")
````
